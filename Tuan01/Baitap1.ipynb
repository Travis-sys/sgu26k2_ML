## Bài 6. Phép toán và nhân ma trận

```python
import torch
a = torch.randn(2, 3)
b = torch.randn(2, 3)
print("a:")
print(a)
print("b:")
print(b)
print("\na+b:")
print(a + b)
print("\na / b:")
print(a / b)
print("\na^2:")
print(a ** 2)
print("\nMatrix multiplication:")
print(a @ b.T)
```

## Bài 7: Các hàm thống kê trên Tensor

```python
import torch
a = torch.randn(3, 4)
print(a)
print("\nMean:")
print(torch.mean(a, dim=0))
print("\nStandard deviation:")
print(torch.std(a, dim=0))
print("\nCumulative sum:")
print(torch.cumsum(a, dim=0))
```

```python
print("\nMean:")
print(torch.mean(a, dim=1))
print("\nStandard deviation:")
print(torch.std(a, dim=1))
print("\nCumulative sum:")
print(torch.cumsum(a, dim=1))
```

#### Yêu cầu:
- `dim=0`: thao tác theo chiều thứ nhất, tức là tính toán dọc theo các hàng. Với `a` có shape `(3, 4)`, kết quả của `mean` và `std` có 4 giá trị, mỗi giá trị tương ứng với một cột. `cumsum` cũng có shape `(3, 4)` và cộng dồn từ hàng trên xuống hàng dưới.
- `dim=1`: thao tác theo chiều thứ hai, tức là tính toán ngang theo từng hàng. Kết quả của `mean` và `std` có 3 giá trị, mỗi giá trị tương ứng với một hàng. `cumsum` cũng có shape `(3, 4)` và cộng dồn từ cột trái sang cột phải.
- Vì vậy, thay `dim=0` bằng `dim=1` làm thay đổi hướng tính toán và kết quả của `mean`, `std`, nhưng không làm thay đổi shape của `cumsum`.

## Bài 8. Autograd cơ bản

```python
import torch

x = torch.tensor(3.6, requires_grad=True)
y = x * x
y.backward()

print("Ham 1: y = x^2")
print("x =", x)
print("y =", y)
print("Dao ham thu cong dy/dx = 2x =", 2 * x.item())
print("x.grad =", x.grad)

## doi ham y=x^2 thanh y = 3*x^2 + 2*x + 1
x_new = torch.tensor(3.6, requires_grad=True)
y_new = 3 * x_new**2 + 2 * x_new + 1
y_new.backward()

print("\nHam 2: y = 3*x^2 + 2*x + 1")
print("x =", x_new)
print("y =", y_new)
print("Dao ham thu cong dy/dx = 6x + 2 =", 6 * x_new.item() + 2)
print("x_new.grad =", x_new.grad)
```

### Giải thích

- `requires_grad = True` yêu cầu PyTorch theo dõi các phép tính liên quan đến x để tính đạo hàm.
- `y = x * x` tạo hàm `y = x^2`.
- `y.backward()` áp dụng quy tắc lan truyền ngược để tính `dy/dx`.
- `x.grad` lưu gradient của `y` theo `x`.

Với `y = x^2`:

- Đạo hàm: `dy/dx =2*x`.
- Tại `x = 3.6`: `dy/dx = 2 * 3.6 = 7.2`.
- Kết quả này giống `x.grad = 7.2`.

Với `y = 3*x^2 + 2*x + 1`:

- Đạo hàm: `dy/dx = 6*x + 2`.
- Tại `x = 3.6`: `dy/dx = 6 * 3.6 + 2 = 23.6`.
- Kết quả này trùng với `x_new.grad = 23.6`.

## Bài 9. Gradient Descent tìm hệ số đa thức

```python
import numpy as np
import torch

polynomial = np.poly1d([1, 2, 3])
N = 20
X = np.random.randn(N, 1) * 5
Y = polynomial(X)
XX = np.hstack([
    X * X,
    X,
    np.ones_like(X)
])

w = torch.randn(3, 1, requires_grad=True)
x = torch.tensor(XX, dtype=torch.float32)
y = torch.tensor(Y, dtype=torch.float32)
optimizer = torch.optim.NAdam([w], lr=0.01)

target_coefficients = torch.tensor([[1.0], [2.0], [3.0]])
print("Initial coefficients:")
print(w.detach().flatten())

for epoch in range(5000):
    optimizer.zero_grad()
    y_pred = x @ w
    mse = torch.mean(torch.square(y - y_pred))
    mse.backward()
    optimizer.step()

print("Learned coefficients [x^2, x, 1]:")
print(w.detach().flatten())
print("Target coefficients:")
print(target_coefficients.flatten())
print("Coefficient errors:")
print((w.detach() - target_coefficients).flatten())
print("Final MSE:", mse.item())
```

### So sánh và giải thích

- `w` gồm ba hệ số theo thứ tự `[x^2, x, 1]`. Vì vậy, nghiệm đúng cần tìm là `[1, 2, 3]`.
- Sau khi huấn luyện, các hệ số học được xấp xỉ `[1.0000, 2.0000, 3.0001]`, rất gần với `[1, 2, 3]`. Sai số nhỏ là do quá trình tối ưu số và số vòng lặp hữu hạn.
- `optimizer.zero_grad()` xóa gradient của vòng lặp trước. Nếu không xóa, gradient mới sẽ bị cộng dồn vào gradient cũ.
- `mse.backward()` dùng Autograd để tính gradient của hàm mất mát MSE theo `w`.
- `optimizer.step()` dùng các gradient vừa tính để cập nhật `w` theo thuật toán NAdam, giúp MSE giảm dần.
- `y_pred = x @ w` dự đoán giá trị của đa thức; `mse` đo độ khác nhau giữa dự đoán và dữ liệu thật.

## Bài 10. Giải hệ phương trình bằng Autograd

```python
import random
import torch

A = torch.tensor(random.random(), requires_grad=True)
B = torch.tensor(random.random(), requires_grad=True)
C = torch.tensor(random.random(), requires_grad=True)
D = torch.tensor(random.random(), requires_grad=True)
EPOCHS = 2000
optimizer = torch.optim.NAdam([A, B, C, D], lr=0.01)

for _ in range(EPOCHS):
    y1 = A + B - 9
    y2 = C - D - 1
    y3 = A + C - 8
    y4 = B - D - 2
    sqerr = y1**2 + y2**2 + y3**2 + y4**2

    optimizer.zero_grad()
    sqerr.backward()
    optimizer.step()

with torch.no_grad():
    residuals = torch.tensor([
        A + B - 9,
        C - D - 1,
        A + C - 8,
        B - D - 2
    ])
    final_sqerr = torch.sum(residuals**2)

print("A =", A.item())
print("B =", B.item())
print("C =", C.item())
print("D =", D.item())
print("\nKiem tra cac phuong trinh:")
print("A + B =", (A + B).item(), "(muc tieu: 9)")
print("C - D =", (C - D).item(), "(muc tieu: 1)")
print("A + C =", (A + C).item(), "(muc tieu: 8)")
print("B - D =", (B - D).item(), "(muc tieu: 2)")
print("Sai so tung phuong trinh:", residuals)
print("Tong sai so binh phuong:", final_sqerr.item())
```

### Nhận xét

- Sau khi tối ưu, bốn phương trình đều được thỏa mãn gần đúng; tổng sai số bình phương rất nhỏ nên nghiệm số là phù hợp.
- Hệ phương trình có nhiều nghiệm vì phương trình thứ tư phụ thuộc vào ba phương trình đầu:
  `B - D = (9 - A) - (7 - A) = 2`.
- Đặt `A = t` tùy ý, ta có họ nghiệm:
  `A = t`, `B = 9 - t`, `C = 8 - t`, `D = 7 - t`.
- Do đó, bài toán có một tham số tự do và không có một nghiệm duy nhất. NAdam chỉ tìm một nghiệm cụ thể tùy thuộc vào giá trị khởi tạo và quá trình tối ưu.

## Bài 11. Tổng hợp Tensor, Autograd và tối ưu đa thức

```python
import torch

# Tao Tensor X co 24 phan tu va reshape theo ba cach
X = torch.arange(24)
X_4x6 = X.reshape(4, 6)
X_2x12 = X.reshape(2, 12)
X_2x3x4 = X.reshape(2, 3, 4)
print("X reshape 4 x 6:")
print(X_4x6)
print("\nX reshape 2 x 12:")
print(X_2x12)
print("\nX reshape 2 x 3 x 4:")
print(X_2x3x4)

# Phep toan va thong ke theo cot
A = torch.randn(3, 4)
B = torch.randn(3, 4)
print("\nA + B:")
print(A + B)
print("\nA * B:")
print(A * B)
print("\nA / B:")
print(A / B)
print("\nTrung binh theo cot:")
print(torch.mean(A, dim=0))
print("\nDo lech chuan theo cot:")
print(torch.std(A, dim=0))

# Autograd cho y = 2*x^2 + 5*x + 3 tai x = 2
x_value = torch.tensor(2.0, requires_grad=True)
y_value = 2 * x_value**2 + 5 * x_value + 3
y_value.backward()
manual_gradient = 4 * 2 + 5
print("\nAutograd:")
print("y =", y_value.item())
print("dy/dx bang Autograd =", x_value.grad.item())
print("dy/dx thu cong =", manual_gradient)

# Hoc lai he so cua y = 2*x^2 - 3*x + 5 bang NAdam
polynomial = torch.tensor([2.0, -3.0, 5.0])
torch.manual_seed(0)
x_train = torch.linspace(-5, 5, 40).reshape(-1, 1)
y_train = polynomial[0] * x_train**2 + polynomial[1] * x_train + polynomial[2]
features = torch.cat([x_train**2, x_train, torch.ones_like(x_train)], dim=1)
coefficients = torch.randn(3, 1, requires_grad=True)
optimizer = torch.optim.NAdam([coefficients], lr=0.01)

for _ in range(5000):
    optimizer.zero_grad()
    prediction = features @ coefficients
    loss = torch.mean((y_train - prediction)**2)
    loss.backward()
    optimizer.step()

print("\nHe so hoc duoc [x^2, x, 1]:")
print(coefficients.detach().flatten())
print("He so muc tieu:")
print(polynomial)
print("Sai so he so:")
print((coefficients.detach().flatten() - polynomial))
print("MSE cuoi:", loss.item())
```

### Kết quả

- `X` có 24 phần tử nên có thể reshape thành `4 x 6`, `2 x 12` hoặc `2 x 3 x 4`; tổng số phần tử vẫn là 24.
- Với `A` và `B` có shape `(3, 4)`, các phép `A + B`, `A * B` và `A / B` đều thực hiện theo từng phần tử và cho shape `(3, 4)`.
- `torch.mean(A, dim=0)` và `torch.std(A, dim=0)` tính theo cột, nên mỗi kết quả có 4 giá trị.
- Với `y = 2*x^2 + 5*x + 3`, đạo hàm là `dy/dx = 4x + 5`. Tại `x = 2`, kết quả thủ công và Autograd đều bằng 13.
- Với đa thức `y = 2*x^2 - 3*x + 5`, các hệ số học được theo thứ tự `[x^2, x, 1]` xấp xỉ `[2, -3, 5]`.
