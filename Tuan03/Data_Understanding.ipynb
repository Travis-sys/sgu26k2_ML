```python
import matplotlib.pyplot as plt

# Dữ liệu Tuổi, bỏ giá trị thiếu ?
tuoi = [25, 30, 22, 40, 35, 29, 50, 27, 31]

# Các khoảng tuổi
bins = [20, 30, 40, 50, 60]

plt.hist(tuoi, bins=bins, edgecolor='black')

plt.title('Histogram phân phối Tuổi')
plt.xlabel('Tuổi')
plt.ylabel('Số lượng mẫu')

plt.xticks([20, 30, 40, 50, 60])

plt.show()
```
