```python
import matplotlib.pyplot as plt

# Dữ liệu Tuổi
tuoi = [25, 30, 22, 40, 35, 29, 50, 27, 31]

# Vẽ Histogram
plt.hist(tuoi, bins=5, edgecolor='black')

plt.title("Histogram - Tuổi")
plt.xlabel("Tuổi")
plt.ylabel("Tần số")

plt.show()
# Q1 (Tứ phân vị thứ nhất): Là giá trị mà tại đó 25% số liệu nhỏ hơn nó và 75% số liệu lớn hơn hoặc bằng nó trong một tập dữ liệu được sắp xếp theo thứ tự tăng dần.
#Q3 (Tứ phân vị thứ ba): Là giá trị mà tại đó 75% số liệu nhỏ hơn nó và 25% số liệu lớn hơn hoặc bằng nó trong tập dữ liệu đã sắp xếp.
#IQR (Khoảng tứ phân vị): Là hiệu số giữa Q3 và Q1: IQR = Q3 - Q1. Đây là một thước đo quan trọng về độ phân tán của dữ liệu xoay quanh trung vị.
```

```python
import matplotlib.pyplot as plt

# Dữ liệu Lương
luong = [8000, 5000, 15000, -2000, 7000, 8500, 30000, 8200, 7800]

# Vẽ Boxplot
plt.boxplot(luong)

plt.title("Boxplot - Lương")
plt.ylabel("Lương")

plt.show()
```

```python
import matplotlib.pyplot as plt

# Dữ liệu Tuổi và Lương
tuoi = [25, 22, 40, 35, 29, 50, 27, 31]
luong = [8000, 5000, 15000, -2000, 8500, 30000, 8200, 7800]

# Vẽ scatter plot
plt.scatter(tuoi, luong)

# Đặt tên biểu đồ
plt.title("Scatter Plot - Tuổi và Lương")

# Đặt tên trục X
plt.xlabel("Tuổi")

# Đặt tên trục Y
plt.ylabel("Lương")

# Hiển thị lưới
plt.grid(True)

# Hiển thị biểu đồ
plt.show()
```

```python
import matplotlib.pyplot as plt

# Dữ liệu tuổi
tuoi = [
    13, 15, 16, 16, 19, 20, 20, 21, 22, 22,
    25, 25, 25, 25, 30, 33, 33, 35, 35, 35,
    35, 36, 40, 45, 46, 52, 70
]

# Vẽ boxplot
plt.boxplot(tuoi)

# Tiêu đề và nhãn
plt.title("Boxplot - Tuổi")
plt.ylabel("Tuổi")

# Hiển thị lưới
plt.grid(axis='y')

# Hiển thị biểu đồ
plt.show()
```
